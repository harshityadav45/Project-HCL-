```python
# 2. Check for missing value and duplicate Employee_ID records.

import pandas as pd

df = pd.read_csv("17_hr_attrition.csv")

# Check missing values
print("Missing Values:")
print(df.isnull().sum())

# Check duplicate Employee_ID
print("\nDuplicate Employee_ID Records:")
print(df["Employee_ID"].duplicated().sum())

# Display duplicate Employee_ID records
print("\nDuplicate Employee_ID Rows:")
print(df[df["Employee_ID"].duplicated(keep=False)])
```

```python
# 3. Clean missing Satisfaction_Score or Monthly_Income values with justification.

import pandas as pd

df = pd.read_csv("17_hr_attrition.csv")

# Check missing values before cleaning
print("Missing values before cleaning:")
print(df[["Satisfaction_Score", "Monthly_Income"]].isnull().sum())

# Fill missing values with median
df["Satisfaction_Score"] = df["Satisfaction_Score"].fillna(df["Satisfaction_Score"].median())

df["Monthly_Income"] = df["Monthly_Income"].fillna(df["Monthly_Income"].median())

# Check missing values after cleaning
print("\nMissing values after cleaning:")
print(df[["Satisfaction_Score", "Monthly_Income"]].isnull().sum())
```

```python
# 4. Calculate total headcount, active employees, and overall attrition rate (%).

import pandas as pd

df = pd.read_csv("17_hr_attrition.csv")

# Total Headcount
total_employees = len(df)

# Active Employees
active_employees = df[df["Attrition"] == "No"].shape[0]

# Employees who left
employees_left = df[df["Attrition"] == "Yes"].shape[0]

# Overall Attrition Rate
attrition_rate = (employees_left / total_employees) * 100

print("Total Headcount:", total_employees)
print("Active Employees:", active_employees)
print("Employees Left:", employees_left)
print("Overall Attrition Rate:", attrition_rate, "%")
```

```python
# 5. Compare attrition rates across departments (e.g., Sales, R,D, Operations, HR).

import pandas as pd

df = pd.read_csv("17_hr_attrition.csv")

# Calculate department-wise attrition rate
attrition_table = pd.crosstab(
    df["Department"],
    df["Attrition"],
    normalize="index"
) * 100

# Display attrition table
print("Department-wise Attrition Rate:")
print(attrition_table)
```

```python
# 6. Compare average monthly income and tenure for employeees who left versus stayed.

import pandas as pd
# Load Dataset
df = pd.read_csv("17_hr_attrition.csv")

# Group employees according to Attrition
camparison = df.groupby("Attrition").agg(
    Average_Monthly_Income=("Monthly_Income","mean"),
    Average_Tenure=("Tenure_Years","mean")
)

# Display Result
print("Average Monthly Income and Tenure Comparison:")
print(camparison)
```

```python
# 7. Analyze attrition rates by overtime status (Yes vs No).

import pandas as pd

# Load Dataset
df = pd.read_csv("17_hr_attrition.csv")

# Remove extra spaces from column names
df.columns = df.columns.str.strip()

# Find Overtime column
overtime_col = None

for col in df.columns:
    if col.lower().replace("_", "").replace(" ", "") in ["overtime", "overtimestatus"]:
        overtime_col = col
        break

# Check Overtime column
if overtime_col is None:
    print("Overtime column not found!")
    print("Available Columns:")
    print(df.columns.tolist())

else:
    # Calculate Attrition Rate
    overtime_analysis = pd.crosstab(
        df[overtime_col],
        df["Attrition"],
        normalize="index"
    ) * 100

    # Display Result
    print("Attrition Rate by Overtime Status:")
    print(overtime_analysis)
```

```python
# 8. Compare average satisfaction scores across jon levels or roles.

import pandas as pd

# Load Dataset
df = pd.read_csv("17_hr_attrition.csv")

# Remove extra spaces from column names
df.columns = df.columns.str.strip()

# Find Job Level or Job Role column
job_col = None

for col in df.columns:
    name = col.lower().replace("_", "").replace(" ", "")

    if name in ["joblevel", "jobrole", "role"]:
        job_col = col
        break

# Check required columns
if job_col is None or "Satisfaction_Score" not in df.columns:

    print("Required column not found!")
    print("Available Columns:")
    print(df.columns.tolist())

else:
    # Calculate Average Satisfaction Score
    satisfaction_analysis = df.groupby(job_col)["Satisfaction_Score"].mean()

    # Display Result
    print("Average Satisfaction Score Across Job Levels/Roles:")
    print(satisfaction_analysis)
```

```python
# 9. Analyze attrition across commute-distance bands.

import pandas as pd

# Load Dataset
df = pd.read_csv("17_hr_attrition.csv")

# Remove extra spaces from column names
df.columns = df.columns.str.strip()

# Find commute distance column
distance_col = None

for col in df.columns:
    name = col.lower().replace("_", "").replace(" ", "").replace("-", "")

    if name in ["commutedistance", "distance"]:
        distance_col = col
        break

if distance_col is None:
    print("Commute Distance column not found!")
    print("Available Columns:")
    print(df.columns.tolist())

else:
    # Create Distance Bands
    df["Distance_Band"] = pd.cut(
        df[distance_col],
        bins=[-1, 5, 10, 15, float("inf")],
        labels=["0-5", "6-10", "11-15", "Above 15"]
    )

    # Calculate Attrition Rate
    distance_analysis = pd.crosstab(
        df["Distance_Band"],
        df["Attrition"],
        normalize="index"
    ) * 100

    # Display Result
    print("Attrition Across Commute Distance Bands:")
    print(distance_analysis)
```

```python
# 10. Rank department by average overtime hours per employee.

import pandas as pd

# Load Dataset
df = pd.read_csv("17_hr_attrition.csv")

# Remove extra spaces from column names
df.columns = df.columns.str.strip()

# Find Department and Overtime Hours columns
overtime_col = None

for col in df.columns:
    name = col.lower().replace("_", "").replace(" ", "")

    if "overtime" in name and "hour" in name:
        overtime_col = col
        break

# Check required columns
if "Department" not in df.columns or overtime_col is None:

    print("Required columns not found!")
    print("Available Columns:")
    print(df.columns.tolist())

else:
    # Convert overtime hours into numeric values
    df[overtime_col] = pd.to_numeric(
        df[overtime_col],
        errors="coerce"
    )

    # Calculate average overtime hours department-wise
    overtime_analysis = df.groupby("Department")[overtime_col].mean()

    # Rank departments
    department_ranking = overtime_analysis.sort_values(
        ascending=False
    )

    # Display Result
    print("Department Ranking by Average Overtime Hours:")
    print(department_ranking)
```

```python
# 11. Use NumPy to calculate mean, median, and standard deviation of monthly income.

import pandas as pd
import numpy as np

# Load Dataset
df = pd.read_csv("17_hr_attrition.csv")

# Get Monthly Income
income = pd.to_numeric(df["Monthly_Income"], errors="coerce")

# Calculate Mean
mean_income = np.nanmean(income)

# Calculate Median
median_income = np.nanmedian(income)

# Calculate Standard Deviation
std_income = np.nanstd(income)

# Display Results
print("Monthly Income Analysis")

print("Mean:", mean_income)

print("Median:", median_income)

print("Standard Deviation:", std_income)
```

```python
# 12. Identify salary outliers within the dataset using the IQR method.

import pandas as pd
import numpy as np

# Load Dataset
df = pd.read_csv("17_hr_attrition.csv")

# Get Monthly Income
income = pd.to_numeric(df["Monthly_Income"], errors="coerce").dropna()

# Calculate Q1 and Q3
Q1 = np.percentile(income, 25)
Q3 = np.percentile(income, 75)

# Calculate IQR
IQR = Q3 - Q1

# Calculate Lower and Upper Limits
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

# Identify Salary Outliers
outliers = df[
    (df["Monthly_Income"] < lower_limit) |
    (df["Monthly_Income"] > upper_limit)
]

# Display Results
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)

print("\nLower Limit:", lower_limit)
print("Upper Limit:", upper_limit)

print("\nSalary Outliers:")
print(outliers[["Employee_ID", "Monthly_Income"]])

print("\nTotal Salary Outliers:", len(outliers))
```

```python
# 13. Create tenure bands (e.g., <2 yrs, 2\text{--}5 yrs,5+ yrs) and compare attrition across them.

import pandas as pd

# Load Dataset
df = pd.read_csv("17_hr_attrition.csv")

# Remove extra spaces from column names
df.columns = df.columns.str.strip()

# Find Tenure Column
tenure_col = None

for col in df.columns:
    name = col.lower().replace("_", "").replace(" ", "")

    if name in ["yearsatcompany", "tenure", "tenureyears"]:
        tenure_col = col
        break

# Check Required Columns
if tenure_col is None or "Attrition" not in df.columns:

    print("Required columns not found!")
    print("Available Columns:")
    print(df.columns.tolist())

else:
    # Convert Tenure into Numeric
    df[tenure_col] = pd.to_numeric(
        df[tenure_col],
        errors="coerce"
    )

    # Create Tenure Bands
    df["Tenure_Band"] = pd.cut(
        df[tenure_col],
        bins=[-float("inf"), 2, 5, float("inf")],
        labels=["Less than 2 years", "2 to 5 years", "5+ years"],
        right=False
    )

    # Compare Attrition Across Tenure Bands
    tenure_analysis = pd.crosstab(
        df["Tenure_Band"],
        df["Attrition"],
        normalize="index"
    ) * 100

    # Display Results
    print("Attrition Across Tenure Bands:")
    print(tenure_analysis)
```

```python
# 14. Create a department summary table combining headcount, average salary, and attrition rate.

import pandas as pd

# Load Dataset
df = pd.read_csv("17_hr_attrition.csv")

# Remove extra spaces from column names
df.columns = df.columns.str.strip()

# Convert Monthly Income into numeric
df["Monthly_Income"] = pd.to_numeric(
    df["Monthly_Income"],
    errors="coerce"
)

# Create Department Summary
department_summary = df.groupby("Department").agg(
    Headcount=("Employee_ID", "count"),
    Average_Salary=("Monthly_Income", "mean"),
    Employees_Left=("Attrition", lambda x: (x == "Yes").sum())
)

# Calculate Attrition Rate
department_summary["Attrition_Rate"] = (
    department_summary["Employees_Left"] /
    department_summary["Headcount"]
) * 100

# Display Final Summary
print("Department Summary Table:")
print(department_summary.round(2))
```
